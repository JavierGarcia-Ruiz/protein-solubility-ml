This notebook examines sequence length distributions, class proportions, and physicochemical properties across DeepSol (Khurana et al., 2018) and eSOL (Niwa et al., 2009).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from solubility.visualization.plotting import set_style, OKABE_ITO

set_style()
df_te = pd.read_parquet('../data/processed/deepsol_test.parquet')
df_esol = pd.read_parquet('../data/processed/esol_test.parquet')
print(f'DeepSol Test shape: {df_te.shape}, Soluble fraction: {df_te["label"].mean():.3f}')
print(f'eSOL Test shape: {df_esol.shape}, Mean solubility: {df_esol["solubility"].mean():.3f}')
```

```python
# Top physicochemical differences between soluble and insoluble proteins
key_feats = ['net_charge_ph7', 'gravy', 'frac_K', 'lys_arg_ratio', 'frac_hydrophobic', 'aliphatic_index']
diff_df = df_te.groupby('label')[key_feats].mean().T
diff_df.columns = ['Insoluble', 'Soluble']
diff_df['Difference'] = diff_df['Soluble'] - diff_df['Insoluble']
diff_df
```
